```python
import plotly.express as px
import plotly.io as pio
import plotly.graph_objects as go

import matplotlib.pyplot as plt

import seaborn as sns
import numpy as np
import pandas as pd
import copy
```

```python
from abc import ABC, abstractmethod

class Problem(ABC):
    @abstractmethod
    def __call__(self, x: np.ndarray) -> float:
        """Compute the function value at point x."""
        raise NotImplementedError

    @abstractmethod
    def grad(self, x: np.ndarray) -> np.ndarray:
        """Compute the gradient at point x."""
        raise NotImplementedError

class Sphere(Problem):
    def __call__(self, x: np.ndarray) -> float:
        return np.sum(x**2)

    def grad(self, x: np.ndarray) -> np.ndarray:
        return 2 * x

class Rosenbrock(Problem):
    def __call__(self, x: np.ndarray) -> float:
        return np.sum(100.0 * (x[1:] - x[:-1]**2)**2 + (1 - x[:-1])**2)

    def grad(self, x: np.ndarray) -> np.ndarray:
        grad = np.zeros_like(x)
        n = x.size
        grad[0] = -400 * x[0] * (x[1] - x[0]**2) - 2 * (1 - x[0])
        for i in range(1, n - 1):
            grad[i] = 200 * (x[i] - x[i-1]**2) - 400 * x[i] * (x[i+1] - x[i]**2) - 2 * (1 - x[i])
        grad[-1] = 200 * (x[-1] - x[-2]**2)
        return grad

class Rastrigin(Problem):
    def __call__(self, x: np.ndarray) -> float:
        A = 10
        n = x.size
        return A * n + np.sum(x**2 - A * np.cos(2 * np.pi * x))

    def grad(self, x: np.ndarray) -> np.ndarray:
        A = 10
        return 2 * x + 2 * np.pi * A * np.sin(2 * np.pi * x)

class Ackley(Problem):
    def __call__(self, x: np.ndarray) -> float:
        x1, x2 = x
        a = 20
        b = 0.2
        c = 2 * np.pi

        term1 = -a * np.exp(-b * np.sqrt(0.5 * (x1**2 + x2**2)))
        term2 = -np.exp(0.5 * (np.cos(c * x1) + np.cos(c * x2)))

        return term1 + term2 + np.e + a

    def grad(self, x: np.ndarray) -> np.ndarray:
        pass
```

```python
def prepare_mesh_grid(
    problem: Problem,
    bounds_x: tuple[float, float] = (-5.0, 5.0),
    bounds_y: tuple[float, float] = (-5.0, 5.0),
    grid_size: int = 50,
) -> tuple[np.ndarray, np.ndarray, np.ndarray, np.ndarray, np.ndarray]:
    x_vals = np.linspace(bounds_x[0], bounds_x[1], grid_size)
    y_vals = np.linspace(bounds_y[0], bounds_y[1], grid_size)
    X, Y = np.meshgrid(x_vals, y_vals)

    Z = np.zeros_like(X)
    for i in range(grid_size):
        for j in range(grid_size):
            xy = np.array([X[i, j], Y[i, j]])
            Z[i, j] = problem(xy)

    return X, Y, Z, x_vals, y_vals
```

```python
def plot_3d_surface(
    problem: Problem,
    grid_size: int = 50,
):
    X, Y, Z, _, _ = prepare_mesh_grid(problem=problem, grid_size=grid_size)

    fig = go.Figure(
        data=go.Surface(
            x=X,
            y=Y,
            z=Z,
            colorscale="Viridis",
            colorbar=dict(title="f(x, y)", thickness=15, len=0.6),
        )
    )

    fig.update_layout(
        title=problem.__class__.__name__,
        width=800,
        height=700,
        scene=dict(
            xaxis_title="x",
            yaxis_title="y",
            zaxis_title="f(x, y)",
            camera=dict(eye=dict(x=1.5, y=1.5, z=1.0)),  # starting camera angle
        ),
    )

    fig.show()
```

```python
def plot_contour_and_paths(
    problem: Problem,
    paths: list[np.ndarray],
    bounds_x: tuple[float, float] = (-5.0, 5.0),
    bounds_y: tuple[float, float] = (-5.0, 5.0),
    grid_size: int = 200,
    title: str = "",
):
    _, _, Z, x_vals, y_vals = prepare_mesh_grid(problem, bounds_x=bounds_x, bounds_y=bounds_y, grid_size=grid_size)

    fig = go.Figure(
        data=go.Contour(
            x=x_vals, y=y_vals, z=Z,
            colorscale="Viridis",
            contours=dict(showlines=False),
            colorbar=dict(title="Function Value"),
        )
    )

    colors = ['red', 'green', 'orange', 'cyan', 'magenta', 'yellow', 'pink', 'brown']

    for idx, path in enumerate(paths):
        color = colors[idx % len(colors)]
        fig.add_trace(go.Scatter(
            x=path[:, 0], y=path[:, 1],
            mode="lines+markers",
            marker=dict(size=3, color=color),
            line=dict(width=2, color=color),
            name=f"Run {idx+1}",
            showlegend=True,
        ))

    min_idx = np.unravel_index(np.argmin(Z), Z.shape)
    fig.add_trace(go.Scatter(
        x=[x_vals[min_idx[1]]], y=[y_vals[min_idx[0]]],
        mode="markers",
        marker=dict(size=12, color="white", symbol="star", line=dict(width=1, color="black")),
        name="Minimum", showlegend=True,
    ))

    if paths:
        x0, y0 = paths[0][0]
        fig.add_trace(go.Scatter(
            x=[x0], y=[y0],
            mode="markers",
            marker=dict(size=8, color="white", symbol="diamond", line=dict(width=1, color="black")),
            name="Initial starting point",
            showlegend=True,
        ))

    fig.update_layout(
        title=title, xaxis_title="x", yaxis_title="y", width=800, height=700,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
    )
    fig.show()
```

```python
def plot_3d_surface_without_grid(
    problem: Problem,
    grid_size: int = 50,
):
    X, Y, Z, _, _ = prepare_mesh_grid(problem=problem, grid_size=grid_size)

    fig = go.Figure(
        data=go.Surface(
            x=X,
            y=Y,
            z=Z,
            colorscale="Viridis",
            colorbar=dict(title="f(x, y)", thickness=15, len=0.6),
        )
    )

    axis_config = dict(
        showgrid=False,
        zeroline=False,
        showline=False,
        showticklabels=False,
        ticks="",
        title="",
        showbackground=False,
        visible=False,
    )

    fig.update_layout(
        title=problem.__class__.__name__,
        width=800,
        height=700,
        scene=dict(
            xaxis=axis_config,
            yaxis=axis_config,
            zaxis=axis_config,
            camera=dict(eye=dict(x=1.5, y=1.5, z=1.0)),  # starting camera angle
        ),
        paper_bgcolor="white",
        margin=dict(l=0, r=0, t=40, b=0),
    )

    fig.show()
```

```python
rastrigin = Rastrigin()
plot_3d_surface(rastrigin)
plot_contour_and_paths(rastrigin, [], grid_size=500)
```

```python
plot_3d_surface_without_grid(rastrigin)
```

```python
ackley = Ackley()
plot_3d_surface_without_grid(ackley, grid_size=500)
```

```python
import torch
import numpy as np

def optimize(problem, start_point, optimizer_class, epochs=1000, **optimizer_kwargs):
    params = torch.tensor(start_point, dtype=torch.float32, requires_grad=True)
    optimizer = optimizer_class([params], **optimizer_kwargs)

    path = [start_point]
    values = [problem(start_point)]

    for _ in range(epochs):
        optimizer.zero_grad()

        # convert tensor to numpy, call your class, convert result back to tensor
        loss = torch.tensor(problem(params.detach().numpy()), dtype=torch.float32)

        # use your analytical gradient instead of autograd
        grad = torch.tensor(problem.grad(params.detach().numpy()), dtype=torch.float32)
        params.grad = grad  # inject gradient manually

        optimizer.step()

        path.append(params.detach().numpy().copy())
        values.append(loss.item())

    return np.array(path), np.array(values)
```

```python
start = np.array([4.0, 2.0])

path_adam, values_adam = optimize(
    rosenbrock, start,
    torch.optim.Adam, epochs=1000, lr=5
)

path_sgd, values_sgd = optimize(
    rosenbrock, start,
    torch.optim.SGD, epochs=10000, lr=0.0001
)

path_momentum, values_momentum = optimize(
    rosenbrock, start,
    torch.optim.SGD, epochs=10000, lr=0.0001, momentum=0.9
)
```

```python
def remove(path):
    indices = np.logspace(
        0,
        np.log10(len(path) - 1),
        num=200
    ).astype(int)
    indices = np.unique(np.concatenate([[0], indices]))
    return path[indices]
```

```python
path_adam_log = remove(path_adam)
path_momentum_log = remove(path_momentum)

plot_contour_and_paths(rosenbrock, [path_adam_log, path_sgd, path_momentum_log], bounds_x=(-20, 20), bounds_y=(-3, 20), grid_size=500)
```
